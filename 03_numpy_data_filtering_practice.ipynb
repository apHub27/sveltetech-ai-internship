```python
import numpy as np
data = np.array([45,np.nan,48,52,900])
cleaned_data =np.nanmean(data)
print("cleaned data:",cleaned_data)
print("-"*40)
no_outlier_data = np.nanmedian(data)
print("median after outlier removal:",no_outlier_data)

#removing the outlier and blank data from the main data
cleaning_data = np.array(data)<100
cleaned_data = data[cleaning_data]
print("cleaned data:",cleaned_data)
```
