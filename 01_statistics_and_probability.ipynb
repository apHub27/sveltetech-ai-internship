```python
import numpy as np
data_a = np.array((10,20,30,40,50))
data_b = np.array((3,4,5,6,7,8))
print("a mean", data_a.mean())
print("b mean", data_b.mean())
print("a std", data_a.std())
print("b std", data_b.std())
```

```python
import numpy as np
query_len = np.array([12,15,14,18,11,13,16,15,12,110])
mean = query_len.mean()
std = query_len.std()
median = np.median(query_len)

print("mean", mean)
print("std", std)
print("median", median)

#calculating z score using numpy

z_scores = (query_len -mean) / std
for original_val, z in zip(query_len, z_scores):
    print(f"Query Length: {original_val:3d}  ->  Z-Score: {z:.4f}")
print("-" * 50)
#-------------------------------------
#filtering the values >2
clean_val = np.abs(z_scores) < 2
cleaned_queries = query_len[clean_val]

print("CLEANED DATASET (Outliers removed):")
print(cleaned_queries)
```

```python
import numpy as np
raw_speeds = np.array((120, np.nan, 140,110,np.nan,130,150,120))
print("------1.before cleaning----")
print("dirty array with blanks:", raw_speeds)
print("hidden blanks?",np.isnan(raw_speeds))
print("-" * 55)

#calculate avg of valid numbers
avg = np.nanmean(raw_speeds)
print(f"avg of valid numbers:{avg}")
print("-" * 55)

#filling the blank spots
cleaned_speeds = raw_speeds.copy()
cleaned_speeds[np.isnan(cleaned_speeds)] = avg
print("------2.after cleaning----")
print("cleaned array with no blanks:",cleaned_speeds)
```
