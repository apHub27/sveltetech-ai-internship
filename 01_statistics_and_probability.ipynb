```python
import numpy as np
data_a = np.array((10,20,30,40,50))
data_b = np.array((3,4,5,6,7,8))
print("a mean", data_a.mean())
print("b mean", data_b.mean())
print("a std", data_a.std())
print("b std", data_b.std())
```

```python
import numpy as np
query_len = np.array([12,15,14,18,11,13,16,15,12,110])
mean = query_len.mean()
std = query_len.std()
median = np.median(query_len)
print("mean", mean)
print("std", std)
print("median", median)
```
