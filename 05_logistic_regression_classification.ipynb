```python
import numpy as np
# login distance discrepancy
dist_x = np.array([0.1,0.5,1.2,4.5,6.8,8.5])
#check for hack => 0 = No, 1 = Yes
hacked_y = np.array([0,0,0,1,1,1])

#sigmoid formula to squash large value
def sigmoid(x):
    return 1/(1+np.exp(-x))
#calculated internal weights from a trained model
# Equation: z = (1.5 * distance) - 3.5
score = (1.5 * dist_x) -3.5
probability = sigmoid(score)
# If probability is greater than 50% (0.5), classify as Hacked (1) else 0
for dist, prob in zip(dist_x,probability):
  decision = 1 if prob > 0.5 else 0
  print(f"Distance: {dist}k km -> fraud Probability: {prob:.4f}, Decision: {decision}")

```
