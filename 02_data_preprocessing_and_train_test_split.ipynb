```python
#min-max normalization using example of no. of bedrooms and sq foot
import numpy as np
sq_foot = np.array([500,1500,2500,5000])
bedrooms =np.array([1,2,3,5])

print("square foot:",sq_foot)
print("bedrooms:",bedrooms)

#finding min and max

sq_foot_min = np.min(sq_foot)
sq_foot_max = np.max(sq_foot)

bedrooms_min = np.min(bedrooms)
bedrooms_max = np.max(bedrooms)

print("min of sq_foot:",sq_foot_min)
print("max of sq_foot:",sq_foot_max)

print("min of bedrooms:",bedrooms_min)
print("max of bedrooms:",bedrooms_max)

#using MIN-MAX formula : (value-min)/(max-min)

scaled_sq_foot = (sq_foot-sq_foot_min)/(sq_foot_max-sq_foot_min)
scaled_bedrooms = (bedrooms-bedrooms_min)/(bedrooms_max-bedrooms_min)

print("scaled sq_foot:",scaled_sq_foot)
print("scaled bedrooms:",scaled_bedrooms)

```

```python
import numpy as np
#using small datasets of text vector
dataset =np.array([10.5, 22.1, 14.3, 35.8, 18.2, 42.0, 27.4, 31.9, 12.1, 20.5])
#shuffle the data
np.random.seed(42)
shuffled_data = dataset.copy()
np.random.shuffle(shuffled_data)

#calculate splitting index (70% train,20% validate,10 % test)
# out of  10 datasets we will use 7 for train , 23 for validate and 1 for test
Train = int(0.7*len(shuffled_data))
validate = int(0.2*len(shuffled_data))
test = int(0.1*len(shuffled_data))
#slice from idx 7 to 9
train_bucket = shuffled_data[0:Train]
validate_bucket = shuffled_data[Train:Train+validate]
test_bucket = shuffled_data[Train+validate:]

print("Training data:",train_bucket)
print("Validated data:",validate_bucket)
print("tested data:",test_bucket)
```
