```python
import numpy as np

# 1. Imagine 5 customer prompts: [Word Count]
prompt_word_counts = np.array([12, 85, 95, 15, 60])

# Labels (y): Should it go to the expensive LLM? (0 = No, 1 = Yes)
expensive_llm_y  = np.array([0,  1,  1,  0,  1])

print("=== RAW PROMPT DISTRIBUTION ===")
print("Word Counts:", prompt_word_counts)
print("LLM Target :", expensive_llm_y)
print("-" * 55)

# 2. THE DECISION SPLIT OPERATION: What happens if the tree asks: "Is Word Count > 50?"
threshold_split = 50

# Create our True/False checklist mask (The conditional node of the tree)
left_branch_mask  = prompt_word_counts > threshold_split  # Greater than 50
right_branch_mask = prompt_word_counts <= threshold_split # Less than or equal to 50

# Extract the answers inside each branch using your masking skill
left_branch_answers  = expensive_llm_y[left_branch_mask]
right_branch_answers = expensive_llm_y[right_branch_mask]

print(f"=== TREE NODE SPLIT RESULT (Threshold: > {threshold_split} words) ===")
print("Left Branch Answers (Prompts > 50) :", left_branch_answers)
print("💡 Look: This bucket is 100% clean! All items are 1. Zero impurity.")
print("\nRight Branch Answers (Prompts <= 50):", right_branch_answers)
print("💡 Look: This bucket is also 100% clean! All items are 0. Zero impurity.")
print("-" * 55)
print("🚀 Result: The computer accepts this split because it perfectly separated the data!")
```
